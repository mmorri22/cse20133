### Basic Recursive Tracing

Open the file <code>trace.c</code> and view its contents.

On the overhead, we will "trace" out the recursion the way the machine performs the operation

    > make trace
    gcc -std=c2x -O2 -g -Wall -Wextra -Wconversion -Wshadow -Werror -lm -c trace.c
    gcc -std=c2x -O2 -g -Wall -Wextra -Wconversion -Wshadow -Werror -lm -o trace trace.o
    ./trace
    In foo: x = 10
    Foo Case 2!
    In bar: y = 9
    Bar Case 3!
    In bar: y = 6
    Bar Case 2!
    In foo: x = 4
    Foo Case 2!
    In bar: y = 3
    Bar Case 3!
    In bar: y = 0
    Bar base case!
    34
    > make clean
    rm -rf *.o *.swp trace trace_ptr fib fib_memo
